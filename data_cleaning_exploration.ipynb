# IntTravel — Nettoyage et Exploration des Données

Ce notebook couvre :

- Nettoyage (valeurs manquantes, doublons, types)

- Export d'un dataset nettoyé pour les étapes suivantes


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', 50)
pd.set_option('display.max_rows', 200)

DATA_PATH = 'dataset/interaction_5.csv'
SAMPLE_ROWS = 200_000  # adjust if needed
```

## 1) Chargement (échantillon)
Le fichier est volumineux. On commence par un échantillon pour l'exploration rapide.


```python
df = pd.read_csv(DATA_PATH, nrows=SAMPLE_ROWS)
df.head()
```

```python
df.info()
```

## 2) Qualité des données
- Valeurs manquantes
- Doublons
- Types de colonnes


```python
missing = df.isna().mean().sort_values(ascending=False)
missing
```

```python
duplicate_count = df.duplicated().sum()
duplicate_count
```

## 3) Colonnes candidates (à adapter)
Selon le dataset, vous pouvez avoir :
- `user_id`
- `poi_id` / `item_id`
- `rating` / `interaction` / `score`
- `timestamp`
- `region`, `transport`, `weather`, etc.

Mettez à jour ces variables si les noms diffèrent.


```python
USER_COL = 'user_id'
ITEM_COL = 'poi_id'
RATING_COL = None  # e.g. 'rating'
TIME_COL = None    # e.g. 'timestamp'

# Check which of these columns exist
df.columns
```

## 4) Nettoyage basique (exemple)
- Retirer les doublons
- Gérer les lignes avec user/poi manquants
- Convertir types si nécessaire


```python
df_clean = df.copy()

# Drop exact duplicate rows
df_clean = df_clean.drop_duplicates()

# Drop rows missing user or item
if USER_COL in df_clean.columns:
    df_clean = df_clean[df_clean[USER_COL].notna()]
if ITEM_COL in df_clean.columns:
    df_clean = df_clean[df_clean[ITEM_COL].notna()]

df_clean.shape
```

## 5) Exploration rapide
- Distribution des interactions par utilisateur
- Distribution des interactions par POI
- Top POI


```python
if USER_COL in df_clean.columns:
    user_counts = df_clean[USER_COL].value_counts()
    user_counts.describe()
```

```python
if ITEM_COL in df_clean.columns:
    item_counts = df_clean[ITEM_COL].value_counts()
    item_counts.describe()
```

```python
if ITEM_COL in df_clean.columns:
    item_counts.head(20)
```

## 6) Visualisations
Ajustez les colonnes selon votre dataset.


```python
if USER_COL in df_clean.columns:
    plt.figure(figsize=(6,4))
    sns.histplot(user_counts, bins=50, log_scale=(True, True))
    plt.title('Distribution des interactions par utilisateur')
    plt.xlabel('Nombre d'interactions (log)')
    plt.ylabel('Nombre d'utilisateurs (log)')
    plt.show()
```

```python
if ITEM_COL in df_clean.columns:
    plt.figure(figsize=(6,4))
    sns.histplot(item_counts, bins=50, log_scale=(True, True))
    plt.title('Distribution des interactions par POI')
    plt.xlabel('Nombre d'interactions (log)')
    plt.ylabel('Nombre de POI (log)')
    plt.show()
```
